```python
def extract_label(filename):
    name = os.path.splitext(filename)[0].upper()
    if name.startswith(("IMG", "DSC", "PXL", "PHOTO", "SCREENSHOT")):
        return None
    match = re.match(r"^[A-Z]($|[_\-\s])", name)
    if match:
        return match.group(0)[0]
    return None

labels = []
for root, dirs, files in os.walk(data_dir):
    for f in files:
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.heic')):
            label = extract_label(f)
            if label:
                labels.append(label)

counts = Counter(labels)

for letter, count in sorted(counts.items()):
    print(f"{letter}: {count}")

print("\nTotal labeled images:", sum(counts.values()))
```

# 4.Prepare the Data


Apply any data transformations and explain what and why

```python
def extract_label(filename):
    name = os.path.splitext(filename)[0].upper()
    if name.startswith("IMG"):
        return None
    match = re.match(r"^[A-Z]($|[_\-\s])", name)
    if match:
        return match.group(0)[0]
    return None

all_files = []
for root, _, files in os.walk(data_dir):
    for f in files:
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.heic')):
            all_files.append(os.path.join(root, f))

labels = [extract_label(os.path.basename(f)) for f in all_files]
df = pd.DataFrame({"filepath": all_files, "label": labels})
print(df["label"].value_counts(dropna=False))
```

```python
register_heif_opener()

data_dir = "letters"  

converted_count = 0
for root, _, files in os.walk(data_dir):
    for f in files:
        if f.lower().endswith(".heic"):
            heic_path = os.path.join(root, f)
            png_path = os.path.splitext(heic_path)[0] + ".png"
            try:
                img = Image.open(heic_path)
                img.save(png_path, "PNG")
                converted_count += 1
                os.remove(heic_path)
            except Exception as e:
                print(f"Failed to convert {heic_path}: {e}")

```

```python
image_paths = []
for root, _, files in os.walk(data_dir):
    for f in files:
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.heic')):
            image_paths.append(os.path.join(root, f))

for i, img_path in enumerate(image_paths):
    folder = os.path.dirname(img_path)
    ext = os.path.splitext(img_path)[1]

    try:
        img = Image.open(img_path)
    except:
        continue

    plt.imshow(img)
    plt.axis("off")
    plt.title(f"{i+1}/{len(image_paths)} — {os.path.basename(img_path)}")
    plt.show(block=False)
    plt.pause(0.1)

    label = input("letter or quit: ").strip()
    plt.close("all")

    if label.lower() == "quit":
        break
    if not label:
        continue

    new_name = f"{label}{ext}"
    new_path = os.path.join(folder, new_name)
    counter = 1
    while os.path.exists(new_path):
        new_name = f"{label}_{counter}{ext}"
        new_path = os.path.join(folder, new_name)
        counter += 1

    os.rename(img_path, new_path)
```

```python
def preprocess_image(path, size=(224, 224)):
    img = cv2.imread(path)
    if img is None:
        return None
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (3, 3), 0)
    eq = cv2.equalizeHist(blur)
    resized = cv2.resize(eq, size)
    normalized = resized / 255.0
    return normalized
    
```

```python
from sklearn import svm
from sklearn.svm import SVC
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
import os
import pandas as pd
import joblib
from homeharvest import scrape_property
import matplotlib.pyplot as plt
```

```python
def infrence(model, house):
    loaded_model, train_columns = joblib.load(model)
    test_house_enc = pd.get_dummies(house)
    test_house_enc = test_house_enc.reindex(columns=train_columns, fill_value=0)
    results = loaded_model.predict(test_house_enc)
    results = f"Predicted Price for the New House: ${results[0]:,.2f}"
    return results
```

```python
test_house = pd.DataFrame({
    "zip_code": [33176],
    "beds": [3.0],
    "full_baths": [2.0],
    "half_baths": [1.0],
    "sqft": [3525],
    "lot_sqft": [0],   
    "neighborhoods": ["Brickell"],
    "stories": [1.0]})
```
